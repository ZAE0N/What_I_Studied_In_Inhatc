```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()
strAPI_key = os.getenv("OPENAI_API_KEY")    # Load OpenAI API Key
client = OpenAI(api_key=strAPI_key)         # Create OpenAI Client Instance 
```

```python
# Create TTS 
response = client.audio.speech.create(
    model="tts-1-hd",
    voice="alloy",
    input="Hello! This is a TTS API test.",
)

response.write_to_file("C:/LLM_Project/Data/Audio/LLM_TTS_Sample.mp3")

# Play the audio file
import IPython.display as ipd

ipd.Audio("C:/LLM_Project/Data/Audio/LLM_TTS_Sample.mp3")
```

```python
# Change voice and output file name
voice = "ash"
mp3_file = f"C:/LLM_Project/Data/Audio/LLM_TTS_Sample_{voice}.mp3"
Source_txtFile = f"C:/LLM_Project/Data/AI_agent.txt"

# Read text from file(C:/LLM_Project/Data/AI_agent.txt)
with open(Source_txtFile, "r", encoding="utf-8") as f:
    srcText_content = f.read()

# Create TTS 
response = client.audio.speech.create(
    model="tts-1-hd",
    voice=voice,                    # Change voice
    input=srcText_content,          # Change input text
)

response.write_to_file(mp3_file)    # Change output file name

# Play the audio file
import IPython.display as ipd

ipd.Audio(mp3_file)
```
